3. Whether a given point belongs to a plane.

## 2. Mathematical Equation of a Plane

Let:

- P₀ = (x₀, y₀, z₀) be a point on the plane.
- n = (a, b, c) be a non-zero normal vector.
- P = (x, y, z) be an arbitrary point.

The vector P - P₀ lies in the plane if and only if
it is perpendicular to the normal vector.

Therefore:

$$
\mathbf{n}\cdot(P-P_0)=0
$$

Expanding the dot product:

$$
a(x-x_0)+b(y-y_0)+c(z-z_0)=0
$$

### Example

Consider a plane defined by:

$$
P_0=(0,0,0)
$$

$$
\mathbf{n}=(0,0,1)
$$

Its equation becomes:

$$
z=0
$$

This is the XY-plane.

The point A = (3, 4, 0) belongs to this plane,
while B = (3, 4, 5) does not.

### Hypothesis

A point belongs to the plane if the dot product
of its displacement vector and the plane's
normal vector equals zero.

```python
from geometry_engine.point import Point3D
from geometry_engine.vector import Vector3D
from geometry_engine.plane import Plane3D

plane = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 1)
)

point_a = Point3D(3, 4, 0)
point_b = Point3D(3, 4, 5)

print("Point A belongs to plane:", plane.contains_point(point_a))
print("Point B belongs to plane:", plane.contains_point(point_b))
```

## 3. Distance from a Point to a Plane

The shortest distance between a point and a plane
is measured along the direction perpendicular to the plane.

Let:

- P be an arbitrary point.
- P₀ be a point on the plane.
- n be a non-zero normal vector.

The distance formula is:

$$
d(P,\Pi)=\frac{|\mathbf{n}\cdot(P-P_0)|}{\|\mathbf{n}\|}
$$

The denominator normalizes the result, making the
distance independent of the magnitude of the normal vector.

### Example

Consider the plane:

$$
z=0
$$

and the point:

$$
P=(3,4,5)
$$

The normal vector is:

$$
\mathbf{n}=(0,0,1)
$$

Therefore:

$$
d(P,\Pi)=\frac{|5|}{1}=5
$$

### Hypothesis

Multiplying the normal vector by a non-zero scalar
does not change the distance from a point to the plane.

```python
point = Point3D(3, 4, 5)

plane_1 = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 1)
)

plane_2 = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 2)
)

print("Distance with unit normal:", plane_1.distance_to_point(point))
print("Distance with scaled normal:", plane_2.distance_to_point(point))
```

## 4. Experimental Results and Conclusions

### Experimental Results

We investigated the distance from the point
P = (3, 4, 5) to the plane z = 0.

Two different normal vectors were used:

- n₁ = (0, 0, 1)
- n₂ = (0, 0, 2)

Both calculations produced the same distance:

$$
d_1 = d_2 = 5
$$

### Conclusion

The experimental results support our hypothesis:
scaling a plane's normal vector by a non-zero
constant does not change the calculated distance.

This property follows from the normalization
factor in the point-to-plane distance formula.

## 5. Intersection of a Line and a Plane

### Mathematical Model

A line in three-dimensional space is defined by:

$$
\mathbf{r}(t)=\mathbf{P}_0+t\mathbf{d}
$$

where:

- P₀ is a point on the line.
- d is the direction vector.
- t is a real parameter.

A plane is defined by:

$$
\mathbf{n}\cdot(\mathbf{r}-\mathbf{Q}_0)=0
$$

where:

- Q₀ is a point on the plane.
- n is the normal vector.

### Derivation

Substitute the line equation into the plane equation:

$$
\mathbf{n}\cdot(\mathbf{P}_0+t\mathbf{d}-\mathbf{Q}_0)=0
$$

Expanding:

$$
\mathbf{n}\cdot(\mathbf{P}_0-\mathbf{Q}_0)
+t(\mathbf{n}\cdot\mathbf{d})=0
$$

Solving for t:

$$
t=\frac{\mathbf{n}\cdot(\mathbf{Q}_0-\mathbf{P}_0)}
{\mathbf{n}\cdot\mathbf{d}}
$$

The intersection point is then:

$$
\boxed{\mathbf{I}=\mathbf{P}_0+t\mathbf{d}}
$$

### Special Cases

1. If the denominator is non-zero, the line intersects
   the plane at exactly one point.

2. If the denominator is zero and the line's reference
   point belongs to the plane, the entire line lies
   in the plane.

3. If the denominator is zero and the reference point
   does not belong to the plane, the line is parallel
   to the plane and there is no intersection.

### Example

Consider the line:

$$
\mathbf{r}(t)=(1,2,5)+t(0,0,-1)
$$

and the plane:

$$
z=0
$$

Substituting into the plane equation:

$$
5-t=0
$$

Therefore:

$$
t=5
$$

The intersection point is:

$$
\boxed{I=(1,2,0)}
$$

```python
from geometry_engine.line import Line3D

line = Line3D(
    Point3D(1, 2, 5),
    Vector3D(0, 0, -1)
)

plane = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 1)
)

intersection = plane.intersect_line(line)

print("Intersection point:", intersection)
print("Point belongs to plane:", plane.contains_point(intersection))
```

### Special Cases and Error Handling

The GeometryEngine3D library distinguishes three
possible relationships between a line and a plane:

1. **Unique intersection:** The line intersects the
   plane at exactly one point. The method returns Point3D.

2. **Parallel line:** The line is parallel to the plane
   without sharing any points. The method returns None.

3. **Line contained in the plane:** Every point on the
   line belongs to the plane. The method raises ValueError
   because infinitely many intersection points exist.

These cases are determined by examining the numerator
and denominator of the intersection parameter equation.

### Numerical Limitations

The current implementation uses an absolute tolerance
of 1e-12 to identify parallel configurations.

Future work will investigate numerical stability
for geometric objects with different coordinate scales.

```python
# Experiment: Three possible line-plane relationships

plane = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 1)
)

# Case 1: Unique intersection
intersecting_line = Line3D(
    Point3D(1, 2, 5),
    Vector3D(0, 0, -1)
)

print("Case 1 - Unique intersection:")
print(plane.intersect_line(intersecting_line))

# Case 2: Parallel line without intersection
parallel_line = Line3D(
    Point3D(1, 2, 5),
    Vector3D(1, 0, 0)
)

print("\nCase 2 - Parallel line:")
print(plane.intersect_line(parallel_line))

# Case 3: Line entirely contained in the plane
contained_line = Line3D(
    Point3D(1, 2, 0),
    Vector3D(1, 0, 0)
)

print("\nCase 3 - Line contained in the plane:")

try:
    print(plane.intersect_line(contained_line))
except ValueError as error:
    print(error)
```

## 7. Geometric Transformations in 3D Space

Geometric transformations modify the position, size,
or orientation of objects in three-dimensional space.

The three fundamental transformations investigated
in this project are:

1. Translation
2. Scaling
3. Rotation

### 7.1 Translation

Translation moves a point by a specified displacement
vector without changing its shape or orientation.

Let the original point be:

$$
P=(x,y,z)
$$

and the displacement vector be:

$$
\mathbf{v}=(a,b,c)
$$

The translated point is:

$$
P'=P+\mathbf{v}
$$

Therefore:

$$
\boxed{P'=(x+a,\ y+b,\ z+c)}
$$

### Hypothesis

Translation preserves the Euclidean distance between
any two points when the same displacement vector
is applied to both points.

### Mathematical Example

Consider:

$$
P=(2,3,4)
$$

and:

$$
\mathbf{v}=(5,-1,2)
$$

The translated point is:

$$
P'=(7,2,6)
$$

The next experiment verifies the translation formula
using the GeometryEngine3D library.

```python
# Experiment: Translation preserves distances

point_a = Point3D(2, 3, 4)
point_b = Point3D(5, 7, 8)

# Translation vector
dx, dy, dz = 5, -1, 2

# Translate both points by the same vector
translated_a = point_a.translate(dx, dy, dz)
translated_b = point_b.translate(dx, dy, dz)

# Calculate distances before and after translation
distance_before = point_a.distance_to(point_b)
distance_after = translated_a.distance_to(translated_b)

print("Original point A:", point_a)
print("Original point B:", point_b)

print("\nTranslated point A:", translated_a)
print("Translated point B:", translated_b)

print("\nDistance before translation:", distance_before)
print("Distance after translation:", distance_after)

print(
    "Distance preserved:",
    abs(distance_before - distance_after) < 1e-9
)
```

### 7.2 Translation: Experimental Results and Conclusion

#### Experimental Results

Two points were translated using the same displacement
vector:

$$
\mathbf{v}=(5,-1,2)
$$

The original points were:

$$
A=(2,3,4), \quad B=(5,7,8)
$$

After translation:

$$
A'=(7,2,6), \quad B'=(10,6,10)
$$

The calculated distances were:

$$
d(A,B)=\sqrt{41}\approx6.403124
$$

$$
d(A',B')=\sqrt{41}\approx6.403124
$$

#### Mathematical Explanation

For any two points A and B and a translation
vector v:

$$
(A+\mathbf{v})-(B+\mathbf{v})=A-B
$$

Therefore:

$$
\boxed{d(A+\mathbf{v},B+\mathbf{v})=d(A,B)}
$$

#### Conclusion

The experimental results support the hypothesis
that translation preserves Euclidean distances.

Translation is an isometric transformation because
it preserves distances between points.

This property is fundamental in computational
geometry, computer graphics, and 3D simulations.
