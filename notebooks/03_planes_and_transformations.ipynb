## 2. Mathematical Equation of a Plane

Let:

- P₀ = (x₀, y₀, z₀) be a point on the plane.
- n = (a, b, c) be a non-zero normal vector.
- P = (x, y, z) be an arbitrary point.

The vector P - P₀ lies in the plane if and only if
it is perpendicular to the normal vector.

Therefore:

$$
\mathbf{n}\cdot(P-P_0)=0
$$

Expanding the dot product:

$$
a(x-x_0)+b(y-y_0)+c(z-z_0)=0
$$

### Example

Consider a plane defined by:

$$
P_0=(0,0,0)
$$

$$
\mathbf{n}=(0,0,1)
$$

Its equation becomes:

$$
z=0
$$

This is the XY-plane.

The point A = (3, 4, 0) belongs to this plane,
while B = (3, 4, 5) does not.

### Hypothesis

A point belongs to the plane if the dot product
of its displacement vector and the plane's
normal vector equals zero.

```python
from geometry_engine.point import Point3D
from geometry_engine.vector import Vector3D
from geometry_engine.plane import Plane3D

plane = Plane3D(
    Point3D(0, 0, 0),
    Vector3D(0, 0, 1)
)

point_a = Point3D(3, 4, 0)
point_b = Point3D(3, 4, 5)

print("Point A belongs to plane:", plane.contains_point(point_a))
print("Point B belongs to plane:", plane.contains_point(point_b))
```
