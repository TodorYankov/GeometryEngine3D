## 2. Parametric Equation of a Line

A line in three-dimensional Euclidean space can be
defined by a point and a non-zero direction vector.

Let:

- P₀ = (x₀, y₀, z₀) be a point on the line.
- d = (a, b, c) be a non-zero direction vector.
- t ∈ ℝ be a parameter.

The vector equation of the line is:

$$
\mathbf{r}(t) = \mathbf{P}_0 + t\mathbf{d}
$$

Its coordinate representation is:

$$
x(t) = x_0 + at
$$

$$
y(t) = y_0 + bt
$$

$$
z(t) = z_0 + ct
$$

### Example

Consider:

$$
P_0 = (1, 2, 3)
$$

$$
\mathbf{d} = (2, 1, -1)
$$

For t = 2:

$$
P(2) = (1+2\cdot2,\;2+2\cdot1,\;3-2)
$$

Therefore:

$$
\boxed{P(2) = (5, 4, 1)}
$$

```python
from geometry_engine.point import Point3D
from geometry_engine.vector import Vector3D
from geometry_engine.line import Line3D

line = Line3D(
    Point3D(1, 2, 3),
    Vector3D(2, 1, -1)
)

for t in [-2, -1, 0, 1, 2]:
    print(f"t = {t:2} -> {line.point_at(t)}")
```

## 3. Triangle Geometry in 3D

A triangle in three-dimensional Euclidean space is
defined by three non-collinear points A, B, and C.

### 3.1 Side Lengths and Perimeter

The length of each side is calculated using the
Euclidean distance formula:

$$
d(A,B)=\sqrt{(x_B-x_A)^2+(y_B-y_A)^2+(z_B-z_A)^2}
$$

The perimeter is:

$$
P = |AB| + |BC| + |CA|
$$

### 3.2 Area Using the Cross Product

We construct two vectors:

$$
\overrightarrow{AB}=B-A
$$

$$
\overrightarrow{AC}=C-A
$$

The magnitude of their cross product equals the
area of the parallelogram formed by these vectors.

Therefore, the area of the triangle is:

$$
S=\frac{1}{2}
\left\|\overrightarrow{AB}\times\overrightarrow{AC}\right\|
$$

### Example

Let:

$$
A=(0,0,0),\quad B=(3,0,0),\quad C=(0,4,0)
$$

Then:

$$
\overrightarrow{AB}=(3,0,0)
$$

$$
\overrightarrow{AC}=(0,4,0)
$$

Their cross product is:

$$
\overrightarrow{AB}\times\overrightarrow{AC}=(0,0,12)
$$

Thus:

$$
S=\frac{1}{2}\sqrt{0^2+0^2+12^2}=6
$$

The side lengths are 3, 4, and 5.

The perimeter is:

$$
P=3+4+5=12
$$

```python
from geometry_engine.triangle import Triangle3D

triangle = Triangle3D(
    Point3D(0, 0, 0),
    Point3D(3, 0, 0),
    Point3D(0, 4, 0),
)

print("Side lengths:", triangle.side_lengths())
print("Perimeter:", triangle.perimeter())
print("Area:", triangle.area())
```

### 3.3 Area Verification Using Heron's Formula

We can calculate the area of a triangle using
Heron's formula, which depends only on its side lengths.

Let a, b, and c be the three side lengths.

The semi-perimeter is:

$$
s = \frac{a+b+c}{2}
$$

Heron's formula is:

$$
S = \sqrt{s(s-a)(s-b)(s-c)}
$$

For a triangle with sides 3, 4, and 5:

$$
s = \frac{3+4+5}{2}=6
$$

Therefore:

$$
S=\sqrt{6(6-3)(6-4)(6-5)}
$$

$$
S=\sqrt{36}=6
$$

**Hypothesis:** The cross-product method and
Heron's formula produce the same area for a
non-degenerate triangle.

```python
import math

a, b, c = triangle.side_lengths()

s = (a + b + c) / 2

area_heron = math.sqrt(
    s * (s - a) * (s - b) * (s - c)
)

area_cross = triangle.area()

print("Cross-product area:", area_cross)
print("Heron's formula area:", area_heron)
print("Absolute difference:", abs(area_cross - area_heron))

assert math.isclose(
    area_cross,
    area_heron,
    rel_tol=1e-9,
    abs_tol=1e-12,
)
```
