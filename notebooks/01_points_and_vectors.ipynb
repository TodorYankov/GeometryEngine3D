# GeometryEngine3D

## 01 — Points and Motion in 3D Space

GeometryEngine3D is a mathematical engine for defining, calculating,
transforming, and visualizing motion trajectories in three-dimensional space.

A moving object is represented by a point:

\[
P = (x, y, z)
\]

where:

- \(x\) represents the position along the X-axis,
- \(y\) represents the position along the Y-axis,
- \(z\) represents the position along the Z-axis.

The point may represent any object whose position changes in 3D space,
such as a robotic end effector, drone, CNC tool, camera, or simulated object.

The goal is to build a general mathematical model rather than a model
for one specific machine.

An industrial robotic arm will be used as one practical application
of the model.

In GeometryEngine3D, a position in space will be represented by the
`Point3D` class.

## Distance Between Two Points in 3D

Let two points in 3D space be:

$$
A = (x_1, y_1, z_1)
$$

and

$$
B = (x_2, y_2, z_2)
$$

The differences between their coordinates are:

$$
\Delta x = x_2 - x_1
$$

$$
\Delta y = y_2 - y_1
$$

$$
\Delta z = z_2 - z_1
$$

Using the Pythagorean theorem in three dimensions, the distance
between the two points is:

$$
d(A,B) =
\sqrt{
(x_2-x_1)^2 +
(y_2-y_1)^2 +
(z_2-z_1)^2
}
$$

```python
from math import sqrt


def distance_3d(point_a, point_b):
    """Calculate the Euclidean distance between two points in 3D space."""
    dx = point_b[0] - point_a[0]
    dy = point_b[1] - point_a[1]
    dz = point_b[2] - point_a[2]

    return sqrt(dx**2 + dy**2 + dz**2)
```
